```python
import balancepy as bp
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import balancepy.models.P18 as P18

# Load the data
filename = 'data/d1_vis1.csv'
height = 1.66
weight = 53

com, stim, time = bp.getdata_legacy(
    filename = filename,
    body_height = height,
    cut_to_cycles = False)

# Create Figure of recorded data (excluding first cycle, which is omitted to avoid transients)
fig = make_subplots(rows=2, cols=1, shared_xaxes=True, subplot_titles=("Stimulus over Time", "Center of Mass over Time"))
# Plot stim over time
fig.add_trace(go.Scatter(x=time, y=stim, mode='lines', name='Stimulus'), row=1, col=1)
# Plot com over time
fig.add_trace(go.Scatter(x=time, y=com, mode='lines', name='Center of Mass'), row=2, col=1)

fig.show()
```

```python
# load data with stimulus and center of mass response cut to cycles
com, stim, time = bp.getdata_legacy(
    filename = filename,
    body_height = height,
    cut_to_cycles = True)

# perform frequency domain analyses
FD = bp.frequency_analysis(stim, com, samplingrate=90, selected_frequencies = 'double_prts', smoothing = bp.logspace_manual_10s)
TD, TD_par = bp.time_domain_analysis(stim, com, samplingrate=90)

fig2 = bp.bode_plot(FD, TD)
fig2.show()
```

```python
# Create a model
model = P18.P18(weight, height)
model.add_experimental_data(stim,com,samplingrate=90, selected_frequencies = 'double_prts', smoothing = bp.logspace_manual_10s)
model.fit()

model.plot()
```
