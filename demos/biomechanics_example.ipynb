```python
import pandas as pd
import numpy as np
import balancepy.biomechanics as bm
import plotly as plt
import plotly.graph_objects as go
```

```python
height = 1.75  # height in meters
weight = 70  # weight in kg

data = pd.read_csv('data/avoc_poc_s5_t3_sham_vrdata.csv')

time = data['time']
stimulus = data['stim_tx']
com = bm.calculate_com_2segmentmodel(data['sho_tz'], data['sho_ty'], data['hip_tz'], data['hip_ty'], height)


# Create subplots
fig = go.Figure()

# Top plot: stimulus across time
fig.add_trace(go.Scatter(x=time, y=stimulus, mode='lines', name='Stimulus', line=dict(color='blue')))

# Bottom plot: com across time
fig.add_trace(go.Scatter(x=time, y=com, mode='lines', name='Center of Mass (COM)', line=dict(color='green'), yaxis='y2'))

# Update layout for subplots
fig.update_layout(
    title="Stimulus and Center of Mass (COM) Across Time",
    xaxis=dict(title="Time (s)"),
    yaxis=dict(title="Stimulus", titlefont=dict(color='blue'), tickfont=dict(color='blue')),
    yaxis2=dict(title="COM", titlefont=dict(color='green'), tickfont=dict(color='green'), overlaying='y', side='right'),
    legend=dict(x=0.5, y=1.1, orientation="h"),
    height=600,
    width=1000
)

fig.show()
```
