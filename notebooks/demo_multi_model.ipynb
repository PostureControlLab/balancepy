```python
import numpy as np
import plotly.express as px
import balancepy as bp
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
from balancepy.model_sim.peterka18 import Peterka18 as P18
from pprint import pprint
from balancepy.model_sim.multi_model import MultiModel as MM
```

```python
# Load the data using the new anaropia preprocessing API

data_exp = []
for k in range(1, 4):
    filename = f'data/d2_vis{k}.csv'
    body_height_m = 1.71
    
    # Load raw CSV data
    raw_data = np.genfromtxt(filename, delimiter=',', names=True)
    time_raw = raw_data['time']
    
    # Extract stimulus and response using anaropia's sr_config
    # SR_LEGACY_AP uses 'stim_pitch' stimulus and COM-based response
    from balancepy.anaropia import (
        SR_LEGACY_AP, AnaropiaPreprocessingConfig, 
        _extract_stimulus, _extract_response, preprocess
    )
    
    stimulus_raw = _extract_stimulus(raw_data, SR_LEGACY_AP)
    response_raw = _extract_response(raw_data, SR_LEGACY_AP, body_height_m=body_height_m)
    
    # Create preprocessing config with desired parameters
    preproc_config = AnaropiaPreprocessingConfig(
        samplingrate_Hz=90,      # target sampling rate (Hz)
        resample=True,           # resample to target rate
        end_time_seconds=220,    # duration for resampling (seconds)
        cut_to_cycles=True,      # cut data to cycles
        cycle_start_samples=20*90,   # discard first n samples
        cycle_length_samples=20*90   # length of each cycle in samples
    )
    
    # Apply preprocessing: resample, filter, cut to cycles
    stim, time = preprocess(stimulus_raw, time_raw, preproc_config)
    com, _ = preprocess(response_raw, time_raw, preproc_config)
    
    # Create sr_data object with processed signals
    data_exp.append(bp.sr_data(90, stim, com, 'prts'))
```

```python
weight = 70  # in kg
body_height_m = 1.75 # in meters

model1 = P18(weight, body_height_m, data_exp[0], ModelName='vis1')
model1.params.update_multimodel_name('W', 'W1')
model2 = P18(weight, body_height_m, data_exp[1], ModelName='vis2')
model2.params.update_multimodel_name('W', 'W2')
model3 = P18(weight, body_height_m, data_exp[2], ModelName='vis3')
model3.params.update_multimodel_name('W', 'W3')

# Initialize MultiModel with a list of models
m = MM([model1, model2, model3])

m.plot_param_table()
```

```python
# fit

m.fit()
```
