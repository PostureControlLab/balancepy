```python
import balancepy as bp
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import balancepy.models.P18 as P18
import pandas as pd

samplingrate = 1000  # Hz

stim = bp.get_mseq(type="prts80", samplingrate=samplingrate)
time = np.arange(0, len(stim) / samplingrate, 1 / samplingrate)

weight = 67
height = 1.75
model = P18.P18(weight, height)
model.samplingrate = samplingrate
params = model.get_params()

# Run time domain simulation
TDsim = model.simulate(stimulus=stim)

# Get frequency response of model
FDsim = model.frequency_response()

bp.bode_plot(FDsim,TDsim,line_name="simulation")
```
